```python
df = pd.read_csv('../data/raw/creditcard.csv')
```

```python
def time_feat(df):
    """
    Простое преобразование признаков
    """
    # Копируем чтобы не менять оригинал
    data = df.copy()
    
    # 1. Time -> Hour (переводим секунды в часы)
    data['Hour'] = data['Time'] // 3600 %  24 # целое количество часов
    
    # Удаляем старые колонки (опционально)
    data = data.drop(['Time'], axis=1)
    
    return data

# Использование:
# df = pd.read_csv('creditcard.csv')
# df_new = prepare_features(df)
# print(df_new.head())
```

```python
import optuna
from xgboost import XGBClassifier
from sklearn.model_selection import cross_val_score
from sklearn.metrics import f1_score

# Твои фичи
features = ['V10', 'V14', 'V17', 'V4', 'V12', 'V7', 'V6', 'Amount']

# Отбираем только эти фичи
X_train_sel = X_train[features]
X_test_sel = X_test[features]

def objective(trial):
    # Предлагаем параметры
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 100, 300),
        'max_depth': trial.suggest_int('max_depth', 3, 8),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'gamma': trial.suggest_float('gamma', 0, 5),
        'reg_alpha': trial.suggest_float('reg_alpha', 0, 2),
        'reg_lambda': trial.suggest_float('reg_lambda', 0, 2),
        'random_state': 42,
        'eval_metric': 'logloss'
    }
    
    model = XGBClassifier(**params)
    
    # 3-fold CV
    scores = cross_val_score(model, X_train_sel, y_train, cv=3, scoring='f1')
    return scores.mean()

# Запуск оптимизации
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=120) 

# Результаты
print("\n🏆 Лучшие параметры:")
for key, value in study.best_params.items():
    print(f"  {key}: {value}")
print(f"\n📊 Лучший F1: {study.best_value:.4f}")

# Финальная модель
best_model = XGBClassifier(**study.best_params, random_state=42)
best_model.fit(X_train_sel, y_train)

# Оценка
from sklearn.metrics import classification_report
y_pred = best_model.predict(X_test_sel)
print("\n📈 Отчет на тесте:")
print(classification_report(y_test, y_pred))
```
